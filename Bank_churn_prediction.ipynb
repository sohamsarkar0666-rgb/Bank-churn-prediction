```python
df = pd.get_dummies(df, columns = ['Geography', 'Gender'])
```

```python
df_f = df[['Geography', 'Gender']]
df_f = pd.get_dummies(df_f)
df_f = df_f.astype('int')
df_f
```

```python
# Drop original columns from main dataframe
df = df.drop(['Geography', 'Gender'], axis=1)

# Concatenate encoded dataframe
df = pd.concat([df, df_f], axis=1)

df.head()
```

```python
df_cleaned = df.copy()
```

```python
df_cleaned
```

```python
from sklearn.preprocessing import StandardScaler
std_col = ['CreditScore','Age', 'Tenure', 'Balance', 'NumOfProducts', 'HasCrCard', 'EstimatedSalary']
scaler = StandardScaler()
df_cleaned[std_col] = scaler.fit_transform(df_cleaned[std_col])
final_df = df_cleaned.copy()
final_df
```

```python
from sklearn.model_selection import train_test_split

x = final_df.drop('Exited', axis = 1)
y = final_df['Exited']
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.30)
```

```python
final_df.head()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
model_dict = {
    'LR' : LogisticRegression(),
    'DT' : DecisionTreeClassifier(),
    'NB' : GaussianNB(),
    'KNN' : KNeighborsClassifier(),
    'SVM' : SVC()
}
result = []

for name,model in model_dict.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    accu = accuracy_score(y_test, y_pred)
    
    result.append({
        'model name' : name,
        'accuracy' : accu
    })
h = pd.DataFrame(result)
h
```
